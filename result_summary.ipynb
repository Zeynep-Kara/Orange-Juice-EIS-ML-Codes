```python
df0 = pd.read_csv("iter_50/mixture/exp_mixture_class_ppm_xm_50_it_dimm_0.csv")
df1 = pd.read_csv("iter_50/mixture/exp_mixture_class_ppm_xm_50_it_dimm_1.csv")
df2 = pd.read_csv("iter_50/mixture/exp_mixture_class_ppm_xm_50_it_dimm_2.csv")
df3 = pd.read_csv("iter_50/mixture/exp_mixture_class_ppm_xm_50_it_dimm_3.csv")

x0_res_mean = exp.get_average_results(df0)
x1_res_mean = exp.get_average_results(df1)
x2_res_mean = exp.get_average_results(df2)
x3_res_mean = exp.get_average_results(df3)

# print("Top 5 results for each X setting:")
# print("X 0:")
# print(x0_res_mean.sort_values('R2', ascending=False).head(3))
# print("X 1:")
# print(x1_res_mean.sort_values('R2', ascending=False).head(3))
# print("X 2:")
# print(x2_res_mean.sort_values('R2', ascending=False).head(3))
# print("X 3:")
# print(x3_res_mean.sort_values('R2', ascending=False).head(3))
# print("X 4:")
# print(x4_res_mean.sort_values('R2', ascending=False).head(3))

x0_res_mean["Label"] = 0
x1_res_mean["Label"] = 1
x2_res_mean["Label"] = 2
x3_res_mean["Label"] = 3

combined_results = pd.concat([x0_res_mean, x1_res_mean, x2_res_mean, x3_res_mean])

avg_results = combined_results.groupby(['DataModel', 'ModelType', 'Model']).agg({
            'Accuracy': 'mean'
        }).reset_index()


print(avg_results.sort_values('Accuracy', ascending=False).head(10))

avg_results.to_csv("res/mixture_class_ppm_xm.csv", index=False)
```
